```python
import requests
```

```python
r = requests.get("https://cataas.com/api/cats",
                 params={"limit": 15}
                 )

# print(f"Content is: {r.content}")
# print(r)

if r.status_code == 200:
    cats_data = r.json()

    for cat in cats_data:
        print(f"Cat ID: {cat["id"]}")

        if cat.get("tags"):
            print(f"Tags: {", ".join(cat["tags"])}")
        else:
            print("Tags: None")
        print("-" * 30)
else:
    print(f"Failed to get data. Status code: {r.status_code}")
```
