```python
import sqlite3
import pandas as pd

# Connect to the database
conn = sqlite3.connect('travel_planner.db')

# Write a master query to pull everything together
# We use LEFT JOINs to ensure we see every city, even if it's missing a specific data point
master_query = """
    SELECT 
        D.DestinationID AS ID,
        D.CityName AS City,
        D.Country,
        CP.BudgetLevel AS Cost,
        WM.AvgTempC AS 'Avg Temp (°C)',
        WM.RainfallMM AS 'Rainfall (mm)',
        A.ActivityName AS Activity,
        DA.Spotlight_Description AS 'Activity Details',
        V.VibeName AS Vibe
    FROM Destinations D
    LEFT JOIN Cost_Profiles CP ON D.DestinationID = CP.DestinationID
    LEFT JOIN Weather_Monthly WM ON D.DestinationID = WM.DestinationID
    LEFT JOIN Destinations_Activities DA ON D.DestinationID = DA.DestinationID
    LEFT JOIN Activities A ON DA.ActivityID = A.ActivityID
    LEFT JOIN Destination_Vibes DV ON D.DestinationID = DV.DestinationID
    LEFT JOIN Travel_Vibes V ON DV.VibeID = V.VibeID
    ORDER BY D.CityName, A.ActivityName, V.VibeName;
"""

# Ask pandas to run the query and convert it into a DataFrame
df_master = pd.read_sql_query(master_query, conn)

# Close the connection
conn.close()

# Display the resulting table! 
# (In a Jupyter Notebook, 'display()' renders a nice interactive HTML table)
display(df_master)
```
