```python
from flask import Flask, request, jsonify
from flask_cors import CORS
import requests
import random

app = Flask(__name__)
CORS(app)

@app.route('/api/process-tags', methods=['POST'])
def process_tags():
    data = request.json
    user_tags = data.get('tags', []) 
    
    
    targets = {str(t).strip().lower() for t in user_tags}
    
    tag_response = requests.get("https://cataas.com/api/tags")
    if tag_response.status_code == 200:
        all_tags = tag_response.json()
        matches = list(targets.intersection(all_tags))
        
        if matches:
            random_index = random.randint(0, len(matches) - 1)
            selected_tag = matches[random_index]
            
            return jsonify({
                "success": True,
                "url": f"https://cataas.com/cat/{selected_tag}",
                "tag": selected_tag
            })
            
    return jsonify({"success": False, "message": "No matching tags found"}), 404

if __name__ == '__main__':

    app.run(debug=True, port=5001, use_reloader=False)
```
