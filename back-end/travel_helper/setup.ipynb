```python
import os
from sqlalchemy import create_engine, Column, Integer, String, ForeignKey, Date
from sqlalchemy.orm import declarative_base, relationship
from datetime import date

db_name = 'travel_planner.db'

#This is to delete the old file and create a new one

if os.path.exists(db_name):
        os.remove(db_name)
        print(f"{db_name} file has been deleted!")
else:
    print("No file to be deleted!")

# Base class for our classes definitions
Base = declarative_base()

# Table 1 Country
class Country(Base):
    __tablename__ = 'Country'
    
    id = Column(Integer, primary_key=True)
    country = Column(String, nullable=False)
    doneym = Column(String, nullable=False)
    capital = Column(String, nullable=False)
    region = Column(String, nullable=False)
    language = Column(String, nullable=False)
    currency = Column(String, nullable=False)
    
    # Relationship between countries and cities
    cities = relationship("Cities", back_populates="linked_country")
    
    def __repr__(self):
        return f"<Country(country='{self.country}', capital='{self.capital}')>"

# Table 2
class Cities(Base):
    __tablename__ = 'Cities'
    
    id = Column(Integer, primary_key=True)
    city = Column(String, nullable=False)
    country_name = Column(String, nullable=False)

    # the foreign key connection to country table id
    country_id = Column(Integer, ForeignKey('Country.id'))
    
    # relationship pointing backward to the country and forward to activities
    linked_country = relationship("Country", back_populates="cities")
    activities = relationship("Activities", back_populates="linked_city")

    def __repr__(self):
        # return f"<City(city='{self.city}', country_id='{self.country.id}')>"
        return f"<City(city='{self.city}', country='{self.country_name}')>"

# Table 3: Activities
class Activities(Base):
    __tablename__ = 'Activities'
    
    id = Column(Integer, primary_key=True)
    activities = Column(String, nullable=False)
    
    city_id = Column(Integer, ForeignKey('Cities.id'))

    linked_city = relationship("Cities", back_populates="activities")
    
    
    def __repr__(self):
        return f"<Activities(activity='{self.activities}', city_id='{self.city_id}')>"


# Create the SQLite engine. echo=True will print the SQL commands generated under the hood.
engine = create_engine('sqlite:///travel_planner.db', echo=True)

# This command actually creates the database file and all the tables defined above
Base.metadata.create_all(engine)

print("\nSuccess: travel_planner.db and all connected tables have been created!")
```
