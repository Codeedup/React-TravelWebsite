This first box will create the database. It creates the database if one doesn't exist and recreates it if it does exist.

```python
import sqlite3
import pandas as pd
import os

DB_PATH = "travel.db"

if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

cursor.execute("PRAGMA foreign_keys = ON;")
```

This next box will create tables with raw SQL

```python
cursor.execute("""
CREATE TABLE countries (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL UNIQUE,
    region TEXT,
    currency_code TEXT,
    main_language TEXT
);
""")

cursor.execute("""
CREATE TABLE destinations (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    country_id INTEGER NOT NULL,
    destination_type TEXT,

    FOREIGN KEY (country_id) REFERENCES countries(id),
    UNIQUE(name, country_id)
);
""")

cursor.execute("""
CREATE TABLE cost_profiles (
    id INTEGER PRIMARY KEY,
    destination_id INTEGER NOT NULL UNIQUE,
    budget_level TEXT,

    FOREIGN KEY (destination_id) REFERENCES destinations(id)
);
""")

cursor.execute("""
CREATE TABLE weather_monthly (
    id INTEGER PRIMARY KEY,
    destination_id INTEGER NOT NULL,
    month INTEGER NOT NULL,
    avg_temp_c REAL,
    rainfall_mm REAL,

    FOREIGN KEY (destination_id) REFERENCES destinations(id),
    UNIQUE(destination_id, month)
);
""")

cursor.execute("""
CREATE TABLE activity_types (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL UNIQUE
);
""")

cursor.execute("""
CREATE TABLE activities (
    id INTEGER PRIMARY KEY,
    destination_id INTEGER NOT NULL,
    activity_type_id INTEGER NOT NULL,
    activity_name TEXT,

    FOREIGN KEY (destination_id) REFERENCES destinations(id),
    FOREIGN KEY (activity_type_id) REFERENCES activity_types(id)
);
""")


cursor.execute("""
CREATE TABLE travel_vibes (
    id INTEGER PRIMARY KEY,
    name TEXT NOT NULL UNIQUE
);
""")

cursor.execute("""
CREATE TABLE destination_vibes (
    destination_id INTEGER NOT NULL,
    vibe_id INTEGER NOT NULL,

    FOREIGN KEY (destination_id) REFERENCES destinations(id),
    FOREIGN KEY (vibe_id) REFERENCES travel_vibes(id),
    PRIMARY KEY (destination_id, vibe_id)
);
""")

conn.commit()
```

Checks that activities link correctly to both destinations and activity types.

```python
query = """
SELECT
    destinations.name AS destination,
    activity_types.name AS activity_type,
    activities.activity_name
FROM activities
JOIN destinations
    ON activities.destination_id = destinations.id
JOIN activity_types
    ON activities.activity_type_id = activity_types.id;
"""

pd.read_sql_query(query, conn)
```

Now we can test some basic filtering: 

```python
query = """
SELECT
    destinations.name AS destination,
    countries.name AS country,
    cost_profiles.budget_level
FROM destinations
JOIN countries
    ON destinations.country_id = countries.id
JOIN cost_profiles
    ON destinations.id = cost_profiles.destination_id
WHERE cost_profiles.budget_level = 'cheap';
"""

pd.read_sql_query(query, conn)
```

Finally a test of a full query that we could use for the front-end:

```python
query = """
SELECT DISTINCT
    destinations.name AS destination,
    countries.name AS country,
    cost_profiles.budget_level,
    weather_monthly.avg_temp_c,
    activity_types.name AS activity_type
FROM destinations
JOIN countries
    ON destinations.country_id = countries.id
JOIN cost_profiles
    ON destinations.id = cost_profiles.destination_id
JOIN weather_monthly
    ON destinations.id = weather_monthly.destination_id
JOIN activities
    ON destinations.id = activities.destination_id
JOIN activity_types
    ON activities.activity_type_id = activity_types.id
WHERE cost_profiles.budget_level = 'cheap'
AND weather_monthly.month = 7
AND weather_monthly.avg_temp_c >= 25;
"""

pd.read_sql_query(query, conn)
```
